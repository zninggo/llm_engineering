```python
# imports
import os
from dotenv import load_dotenv
from openai import OpenAI

from rich.jupyter import display
from IPython.display import Markdown, display, update_display
```

```python
load_dotenv(override=True)
```

```python
# constants

MODEL_GPT = 'gpt-5.2-codex'
MODEL_LLAMA = 'llama3.2'
```

```python
api_key = os.getenv('OPENAI_API_KEY')
base_url = os.getenv('OPENAI_API_URL')

print(api_key)
print(base_url)
```

```python
# set up environment

class Model:

    def __init__(self):

        self.__client = OpenAI(base_url=os.getenv("OPENAI_API_URL"), api_key=os.getenv("OPENAI_API_KEY"))

    def __stream_prompt(self, ):
        pass

    @property
    def get_client(self):
        return self.__client

    def prompt(self, question, model='gpt-5.2-codex', stream=False):
        system_prompt = """
        你的性格是一个尖酸刻薄的人。
        同时是资深的coder，用不耐烦地态度为实习生解决疑难杂症。
        """
        #

        response = self.__client.responses.create(
            model=model,
            stream=stream,

            input=[
                {'role': 'system', 'content': system_prompt},
                {'role': 'user', 'content': question},
            ],
        )

        result = ''
        markdown_handler = display(Markdown(''), display_id=True)

        if stream:
            for chunk in response:
                if chunk.type == 'response.output_text.delta':
                    result += chunk.delta or ''
                    update_display(Markdown(result), display_id=markdown_handler.display_id)
```

```python
# here is the question; type over this to ask something new

question = """
Please explain what this code does and why:
yield from {book.get("author") for book in books if book.get("author")}
"""
```

```python
# Get gpt-4o-mini to answer, with streaming
client = Model()

client.prompt(question, model=MODEL_GPT, stream=True)
```
